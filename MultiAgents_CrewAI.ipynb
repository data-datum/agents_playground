# Multiagentes con CrewAI - Data Science Team

### ¿Qué son los Multi-Agentes?
Los sistemas multi-agente son arquitecturas de software donde múltiples agentes autónomos e inteligentes colaboran o compiten para lograr un objetivo común o individual. Cada agente tiene sus propios roles, objetivos, capacidades y base de conocimientos, y se comunican entre sí para intercambiar información, coordinar acciones y resolver problemas complejos que un solo agente no podría abordar de manera eficiente.

### ¿Qué es CrewAI?
CrewAI es un framework para construir sistemas de agentes de IA con roles bien definidos, objetivos y herramientas, que pueden orquestar flujos de trabajo colaborativos. Permite definir 'crews' de agentes que trabajan juntos en tareas, simulando un equipo donde cada miembro contribuye con su especialización. CrewAI facilita la asignación de tareas, la comunicación entre agentes y la integración de modelos de lenguaje grandes (LLM) y herramientas externas para lograr resultados complejos de manera estructurada.

## Creamos nuestros agentes

### Creación de Agentes
Se definen cuatro agentes con roles, objetivos y trasfondos específicos: `data_analyst`, `ml_engineer`, `critic` y `reporter`. Cada agente está configurado con el `llm` definido anteriormente y `verbose=True` para mostrar su proceso de pensamiento durante la ejecución. Esto establece la arquitectura del equipo multi-agente.

```python
researcher_llm = LLM(
    model="openrouter/cohere/north-mini-code:free",
    api_key=os.environ["OPENROUTER_KEY"]
)

analyst_llm = LLM(
    model="openrouter/cohere/north-mini-code:free",
    api_key=os.environ["OPENROUTER_KEY"]
)

ml_llm = LLM(
    model="openrouter/cohere/north-mini-code:free",
    api_key=os.environ["OPENROUTER_KEY"]
)

critic_llm = LLM(
    model="openrouter/cohere/north-mini-code:free",
    api_key=os.environ["OPENROUTER_KEY"]
)

reporter_llm = LLM(
    model="openrouter/cohere/north-mini-code:free",
    api_key=os.environ["OPENROUTER_KEY"]
)
```

```python
from crewai import Agent

data_analyst = Agent(
    role="Senior Data Analyst",

    goal="""
    Analyze datasets, identify patterns, detect data quality problems,
    and generate meaningful statistical insights.
    """,

    backstory="""
    You are an experienced data analyst specialized in exploratory
    data analysis and statistical reasoning. You are careful about
    assumptions and never make claims that are not supported by data.
    """,

    llm=analyst_llm,
    verbose=True
)
```

```python
ml_engineer = Agent(
    role="Machine Learning Engineer",

    goal="""
    Develop appropriate machine learning approaches based on the
    findings of the data analyst.
    """,

    backstory="""
    You are an experienced ML engineer. You understand model selection,
    validation, feature engineering, metrics and data leakage.
    You prioritize robust evaluation over impressive-looking results.
    """,

    llm=ml_llm,
    verbose=True
)
```

```python
critic = Agent(
    role="Machine Learning Critic",

    goal="""
    Critically evaluate the analysis and machine learning conclusions.
    Identify methodological errors, unsupported claims, data leakage,
    inappropriate metrics and potential biases.
    """,

    backstory="""
    You are a rigorous reviewer of data science projects. Your job is
    not to produce new models but to find weaknesses in the work of
    other agents.
    """,

    llm=critic_llm,
    verbose=True
)
```

```python
reporter = Agent(
    role="Data Science Technical Writer",

    goal="""
    Produce a clear technical report summarizing the analysis,
    modeling results, limitations and recommendations.
    """,

    backstory="""
    You specialize in communicating complex data science results
    clearly to both technical and non-technical audiences.
    """,

    llm=reporter_llm,
    verbose=True
)
```

### Tareas

```python
from crewai import Task

analysis_task = Task(
    description="""
    Analyze the customer churn dataset.

    Identify:
    - number of observations
    - variables and data types
    - missing values
    - distributions
    - potential outliers
    - relationships between variables
    - possible predictors of churn

    Provide a structured analysis with evidence-based conclusions.
    """,

    expected_output="""
    A structured exploratory data analysis containing:
    1. Data quality assessment
    2. Important statistical patterns
    3. Potential predictors of churn
    4. Important limitations
    """,

    agent=data_analyst
)
```

```python
ml_task = Task(
    description="""
    Based on the data analyst's findings, propose a machine learning
    strategy for predicting customer churn.

    Discuss:
    - target variable
    - preprocessing
    - feature engineering
    - baseline model
    - candidate models
    - validation strategy
    - appropriate metrics
    - potential data leakage
    """,

    expected_output="""
    A machine learning strategy including model selection,
    validation methodology and evaluation metrics.
    """,

    agent=ml_engineer
)
```

```python
critic_task = Task(
    description="""
    Review the data analysis and machine learning strategy.

    Look specifically for:
    - incorrect statistical reasoning
    - inappropriate metrics
    - data leakage
    - overfitting
    - unsupported conclusions
    - missing considerations
    - potential biases

    Clearly distinguish between valid conclusions and claims that
    require additional evidence.
    """,

    expected_output="""
    A critical review containing identified problems,
    their severity and concrete suggestions for improvement.
    """,

    agent=critic
)
```

```python
report_task = Task(
    description="""
    Produce a final Data Science report combining the analysis,
    machine learning strategy and critical review.

    The report should include:
    - Executive summary
    - Data analysis
    - Modeling strategy
    - Evaluation methodology
    - Risks and limitations
    - Recommendations
    """,

    expected_output="""
    A professional technical report suitable for presenting
    to a data science team.
    """,

    agent=reporter
)
```

### Definición de Tareas
Para cada agente, se crea una `Task` (tarea) específica. Cada tarea tiene una descripción detallada de lo que el agente debe hacer y un `expected_output` (resultado esperado). Estas tareas guían el trabajo de cada agente y se asignan a ellos, asegurando que cada uno contribuya con su especialidad al objetivo general del proyecto de Data Science.

```python
from crewai import Crew, Process

crew = Crew(
    agents=[
        data_analyst,
        ml_engineer,
        critic,
        reporter
    ],

    tasks=[
        analysis_task,
        ml_task,
        critic_task,
        report_task
    ],

    process=Process.sequential,
    verbose=True
)
```

### Configuración de la Crew (Equipo)
En este paso, se crea la `Crew` (equipo) de agentes. Se le pasan la lista de `agents` y la lista de `tasks` definidas previamente. Se especifica `process=Process.sequential`, lo que significa que las tareas se ejecutarán una tras otra en el orden en que fueron definidas. `verbose=True` proporciona una salida detallada de la ejecución de la crew.

### Inicio de la Ejecución de la Crew
Finalmente, `await crew.kickoff_async()` inicia la ejecución de todo el proceso multi-agente de forma asíncrona. La crew orquesta a los agentes para que realicen sus tareas en secuencia, compartiendo el contexto y la información necesaria para producir el resultado final. El resultado de todo el proceso se almacena en la variable `result`.

```python
result = await crew.kickoff_async()
```

En este proyecto implementamos un sistema multi-agente utilizando CrewAI, en el que distintos agentes especializados colaboran para resolver una tarea de Data Science.

En lugar de delegar todo el problema a un único LLM, dividimos el proceso en diferentes roles:
* un Data Analyst encargado de explorar y comprender los datos,
* un Machine Learning Engineer responsable de proponer la estrategia de modelado,
* un Critic encargado de evaluar la calidad y consistencia del trabajo realizado, y
* un Technical Writer que integra los resultados en un informe final.

CrewAI permite definir estos agentes, sus objetivos, conocimientos, herramientas y tareas, y establecer cómo se comparte el contexto entre ellos. Además, incorporamos herramientas que permiten a los agentes interactuar con datos y utilizar capacidades externas al modelo.


Finalmente, utilizamos la ejecución asíncrona de CrewAI (kickoff_async) para ejecutar el sistema correctamente dentro del entorno de Jupyter/Colab. De esta manera, el proyecto muestra cómo pasar de un agente individual basado en un ciclo de razonamiento y uso de herramientas a una arquitectura multi-agente en la que diferentes agentes especializados colaboran, intercambian información y revisan los resultados de otros agentes antes de producir una respuesta final.
